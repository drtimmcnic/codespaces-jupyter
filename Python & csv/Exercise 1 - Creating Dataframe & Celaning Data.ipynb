Have to Clean data first

```python
#Option 1. Use 'dropna' method to drop values in each row.
#Problem in slose al lot of data
df_nonNAN = df.dropna()
df_nonNAN.head(3)
```

```python
#Option 2 , use 'fillna' method replace missing data with another value
#Fillna (0) =replaces with 0

filled_df = df.fillna(0)
filled_df.head(3)
```

```python
#Option 3, use 'fill(mean)' method with the AVERAGE (mean) or middle (median).
#Thsi is one column exmaple with total Revenue.
filled_df ['Total Revenue'] = df['Total Revenue'].fillna(df['Total Revenue'].mean())
filled_df.head(3)
```

```python
#Option 4 is to fill all null VALUES with the last value.
fill_last = df.ffill()
fill_last.head(3)
```

```python
#to see the stats of the columns - only floating or int are displayed.
filled_df.describe()
```

```python
# change all floating numbers to decimal to 2 places.
pd.options.display.float_format = '{:.2f}'.format
filled_df.head(1)
```

```python
#convert floating point numbers into Integers.  
filled_df['Units Sold'] = filled_df['Units Sold'].astype(int)
filled_df['Order ID'] = filled_df['Order ID'].astype(int)

#to display the datatypes
display(filled_df.dtypes)
```

```python
filled_df.head(2)
```

```python
#creat a calculated column called 'Total Profit'
filled_df['Total Profit'] = filled_df['Total Revenue'] - filled_df['Total Cost']
filled_df.head(1)
```

```python
#To find the total profit for all the data.
Total_Profit = filled_df['Total Profit'].sum()
print('Total Profit for all the Regions is:', Total_Profit)
```
