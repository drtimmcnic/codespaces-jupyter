```python
#Imports the panda library (which is the main library used for Data analytics)
import pandas as pd
```

```python
# 1. Create a sample DataFrame & assigns it 'df'
data = {'name': ['Alice', 'Bob', 'Charlie'], 'salary': [5000, 12000, 30000]}
df = pd.DataFrame(data)
df
```

```python
# 2. Define a simple function using 'def'
def categorize_salary(x):
    if x <= 8000:
        return 'Standard'
    else:
        return 'High'
```

```python
# 3. Apply the function to a column
df['category'] = df['salary'].apply(categorize_salary)

print(df)
```
